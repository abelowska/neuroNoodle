# Visualisation of unfolded EEG signal

Imports

```python
import os
import sys
sys.path.insert(1, '../')
from unfold import unfold_utils
%matplotlib inline
```

Constants

```python
sampling_rate = 64
```

## Full model

Read file names

```python
unfolded_eeg_data_dir = '../data/unfold_no_ssd_output'

pickle_events_files = sorted([f for f in os.listdir(unfolded_eeg_data_dir) if f.endswith('.pickle')])
file_names = [os.path.splitext(f)[0] for f in pickle_events_files]

print(f'Found: {len(file_names)} pickle files')
```

Read pickle files and transform them into evoked dicts

```python
results_list = []
for file_name in file_names:
    try:
        # read events results
        person_unfolded = unfold_utils.ResultsContainer.read_from_file(f'{unfolded_eeg_data_dir}/{file_name}')
        results_list.append(person_unfolded)
    except Exception as e:
        print(e)

print(f"Read: {len(results_list)} results files")
```

## Visualise results

```python
unfold_utils.plot_results_containers(results_list)
```

```python
person_unfolded.num_parameters
```
