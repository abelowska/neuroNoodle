# Visualisation of unfolded EEG signal

Imports

```python
import os
import pickle
from unfold_utils import *
%matplotlib inline
```

Constants

```python
sampling_rate = 64
```

## Read pickled unfolded EEG files

Read file names

```python
unfolded_eeg_data_dir = '../data/unfold_output'

pickle_events_files = sorted([f for f in os.listdir(unfolded_eeg_data_dir) if f.endswith('.pickle')])
file_names = [os.path.splitext(f)[0] for f in pickle_events_files]

print(f'Found: {len(file_names)} pickle files')
```

Read pickle files and transform them into evoked dicts

```python
unfolded_evokeds = {}

for file_name in file_names:
    try:
        # read events dict
        with open(f'{unfolded_eeg_data_dir}/{file_name}.pickle', 'rb') as handle:
            person_unfolded = pickle.load(handle)
            
        person_unfolded_evokeds = create_evokeds_from_container(
            results_obj=person_unfolded, 
            sampling_rate=sampling_rate,
            montage='biosemi32'
        )
        person_id = file_name.rsplit('_', 1)[0]
        unfolded_evokeds[person_id] = person_unfolded_evokeds
    except Exception as e:
        print(e)
        
print(f'Read: {len(unfolded_evokeds)} results files')

# add averaged results
unfold_results_dict_averages = average_evoked_across_ids(unfolded_evokeds)
unfolded_evokeds.update(unfold_results_dict_averages)
```

## Visualise results

```python
plot_results_dict(unfolded_evokeds)
```
